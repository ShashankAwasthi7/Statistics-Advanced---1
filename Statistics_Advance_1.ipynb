Q.1 : What is a random variable in probability theory?

     —> In probability theory, a random variable is a variable whose value is a numerical outcome of a random phenomenon or experiment. It's a real-valued function that maps each outcome in the sample space (the set of all possible results) of a random experiment to a specific number. Random variables are typically denoted by capital letters like X or Y.


Q.2  : What are the types of random variables?

     →  Random variables are categorized into two main types: discrete and continuous. A discrete random variable has a countable number of possible values, such as the outcome of a dice roll or the number of heads in coin flips. A continuous random variable can take on any value within a given range, like the height of a person or the time it takes for a lightbulb to burn out.


Q.3 : Explain the difference between discrete and continuous distributions.

     → Discrete distributions involve random variables that can only take on distinct, separate, and countable values (like the number of heads in three coin flips), with probabilities assigned to each specific outcome. In contrast, continuous distributions are for variables that can take any value within a given range (like height or temperature), which are uncountably infinite, and probabilities are described by a probability density function (PDF) over the continuous interval.


Q.4 : What is a binomial distribution, and how is it used in probability?

     → A binomial distribution is a probability distribution that calculates the probability of achieving a specific number of "successes" in a fixed number of independent trials, where each trial has only two possible outcomes (success or failure). It's used in probability to model phenomena like the number of heads in coin flips or defective items in a batch, by providing a formula to find the exact probability for a given number of successes (x) over a set number of trials (n) with a constant probability of success (p).

It is important because it provides a standardized way to analyze data: any normal distribution can be converted to the standard normal distribution using z-scores, allowing for the comparison of different datasets, estimation of probabilities, and performance of hypothesis tests.

     Why is it Important?

     Comparison of Datasets: It standardizes different normal distributions, enabling the comparison of data with varying means and standard deviations. For example, you can compare the performance of a student on a test with a different average score and spread than another student.

     Probability Estimation: By converting observations to z-scores, you can use standard normal distribution tables (or software) to find the probability of a value falling above or below a certain point in the distribution.

     Hypothesis Testing: Hypothesis testing, including the z-test, relies on the standard normal distribution to determine if a sample mean is significantly different from a population mean.

     Foundation for Inferential Statistics: The Central Limit Theorem states that the distribution of sample means will approach a normal distribution as sample size increases, making the standard normal distribution a fundamental tool for inferential statistics, according to Homework.Study.com.

     Normalization: It allows for the normalization of data before further statistical analysis.


Q.6 : What is the Central Limit Theorem (CLT), and why is it critical in statistics?

     →  The Central Limit Theorem (CLT) states that the distribution of sample means will approach a normal (bell-shaped) distribution as the sample size increases, regardless of the shape of the original population distribution. It is critical in statistics because it allows us to make inferences about a population based on a sample, justifying the use of the normal distribution in many statistical procedures even when the underlying data are not normally distributed. This enables statistical hypothesis testing, confidence interval estimation, and other powerful analytical tools.  

     Why is the CLT Critical in Statistics?

     Justifies the Normal Distribution: Many statistical techniques, such as hypothesis testing and confidence interval estimation, assume a normal distribution. The CLT shows that even if the original data is not normal, the distribution of sample means will be, allowing these methods to be applied.

     Enables Statistical Inference: The CLT provides a framework for generalizing findings from a sample to an entire population. We can use the properties of the normal distribution to understand population parameters, even when studying the entire population is impossible or impractical.

     Simplifies Analysis: It provides a powerful shortcut for analyzing complex data by allowing us to treat the distribution of sample means as normal, simplifying calculations and making statistical analysis more straightforward.

     Foundation for Many Statistical Techniques: It serves as a fundamental concept that underpins a vast number of statistical procedures used in fields like finance, medicine, social sciences, and quality control.


Q.7  : What is the significance of confidence intervals in statistical analysis?

     → Confidence intervals provide a range of values likely to contain the true population parameter, quantifying the uncertainty of a sample estimate and assessing its precision. They indicate whether a result is likely due to chance, and a narrower interval suggests a more stable and reliable estimate. In practice, confidence intervals offer a more dependable alternative to p-values for understanding the magnitude and clinical usefulness of an effect, supporting more informed data-driven decisions in various fields like medicine and business.  


Q.8 :  What is the concept of expected value in a probability distribution?

     → The concept of expected value is the long-term average outcome of a random variable over many repetitions of an experiment, representing a weighted average of all possible outcomes. It is calculated by multiplying each possible value of the random variable by its corresponding probability and summing these products. The expected value provides a single numerical summary of a probability distribution, indicating the "center of mass" or average tendency of the distribution.


Q.9 : Write a Python program to generate 1000 random numbers from a normal distribution with mean = 50 and standard deviation = 5. Compute its mean and standard deviation using NumPy, and draw a histogram to visualize the distribution.

    

```python
import numpy as np
import matplotlib.pyplot as plt


mean = 50
std_dev = 5
num_samples = 1000
random_numbers = np.random.normal(loc=mean, scale=std_dev, size=num_samples)


computed_mean = np.mean(random_numbers)
computed_std_dev = np.std(random_numbers)


print(f"Generated {num_samples} random numbers from a normal distribution.")
print(f"Theoretical Mean: {mean}, Theoretical Standard Deviation: {std_dev}")
print(f"Computed Mean: {computed_mean:.2f}")
print(f"Computed Standard Deviation: {computed_std_dev:.2f}")


plt.hist(random_numbers, bins=30, edgecolor='black', alpha=0.7)
plt.title('Histogram of Random Numbers from Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

Question 10: You are working as a data analyst for a retail company. The company has
collected daily sales data for 2 years and wants you to identify the overall sales trend.

daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
 235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

● Explain how you would apply the Central Limit Theorem to estimate the average sales
with a 95% confidence interval.

● Write the Python code to compute the mean sales and its confidence interval.

```python
import numpy as np
import scipy.stats as stats


daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]


data = np.array(daily_sales)


n = len(data)
mean = np.mean(data)
std_dev = np.std(data, ddof=1)


confidence = 0.95
alpha = 1 - confidence
t_value = stats.t.ppf(1 - alpha/2, df=n-1)


margin_of_error = t_value * (std_dev / np.sqrt(n))
ci_lower = mean - margin_of_error
ci_upper = mean + margin_of_error


print(f"Mean Sales: {mean:.2f}")
print(f"95% Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")
```
